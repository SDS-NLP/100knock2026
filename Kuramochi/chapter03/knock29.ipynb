# 29. 国旗画像のURLを取得する
テンプレートの内容を利用し、国旗画像のURLを取得せよ。（ヒント: MediaWiki APIのimageinfoを呼び出して、ファイル参照をURLに変換すればよい）

```python
import requests
from knock20 import extract_uk_text
from knock25 import extract_infobox
import json

UK_TEXT  = extract_uk_text()
infobox  = extract_infobox(UK_TEXT)

print(infobox["国旗画像"])
```

```python
file_name = infobox['国旗画像'].strip()
file_name = file_name.replace(' ', '_')
print(file_name)
```

```python
# 宛先
url       = "https://ja.wikipedia.org/w/api.php"

# 注文書（ブラックボックス）
params = {
    "action": "query",        
    "format": "json",              # 受け取り形式
    "prop"  : "imageinfo",         # 欲しい情報（画像情報を取得）
    "titles": f"File:{file_name}", # 画像のタイトル（ファイル名）を指定
    "iiprop": "url"                # 画像のURLを含む情報を要求
}

# headers
headers  = {'User-Agent': 'NLP100Knock_Bot/1.0'}

# 宛先(url)に注文書(palams)とヘッダー(headers)を送る
response = requests.get(url, params=params, headers=headers)

# 辞書型に変換
data     = response.json()

print(json.dumps(data, indent=4, ensure_ascii=False))
```

```python
image_url = data['query']['pages']['-1']['imageinfo'][0]['url']
print(image_url)
```
